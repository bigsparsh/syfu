```python
# llm = ChatGoogleGenerativeAI(
#     model='gemini-2.0-flash',
#     temperature=0
# )
llm = ChatGroq(
    model="llama-3.1-8b-instant",
    temperature=0
)
embedding = NVIDIAEmbeddings(
  model="nvidia/nemotron-3-embed-1b"
)

# llm = ChatNVIDIA(
#     model="deepseek-ai/deepseek-v4-flash",
#     temperature=1,
#     top_p=0.95,
# )

# llm = ChatOllama(
#     model='gemma4:e2b',
#     temperature=0,
# )

res = llm.invoke("Which is the biggest whale in the world? Answer in one word")
print(res)
res = embedding.embed_query("Which is the biggest whale in the world?")
print(res)
```

```python
from langchain_community.document_loaders import (TextLoader, PyPDFLoader)
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_experimental.text_splitter import SemanticChunker
# loader = TextLoader("./docs/memory.md")
# soul = loader.load()[0].page_content
# print(soul)

loader = PyPDFLoader("./docs/harry-st.pdf")
harry = loader.load()


chars = []
for doc in harry:
    if doc.page_content: chars.append(len(doc.page_content))

avg_chars =  sum(chars) // len(chars)
chunk_size = int(avg_chars * 0.25)
chunk_overlap = int(chunk_size * 0.1)

print(f"{avg_chars}, {chunk_size}, {chunk_overlap}")
```

```python
sm_splitter = SemanticChunker(embeddings=embedding)
rts = RecursiveCharacterTextSplitter(separators=[ " "], chunk_size=chunk_size, chunk_overlap=chunk_overlap)
# splitted_harry = rts.split_documents(harry)

splitted_harry = sm_splitter.split_documents(harry)
splitted_harry = [x for x in splitted_harry if (x.page_content).strip() != '']
# doc_meta = []
# for doc in splitted_harry:
#     id = uuid.uuid4()
#     doc_meta.append(id)
#     doc.metadata['id'] = id
```

```python
wie = [len(x.page_content.split(" ")) for x in splitted_harry]
avg_wie = sum(wie) // len(wie)
print("Average words in each chunk:", avg_wie)
wie
```

```python
# collection.add(
#     ids=[str(uuid.uuid4()) for _ in splitted_harry],
#     documents=[x.page_content for x in splitted_harry]
# )
vectorstore = Chroma.from_documents(collection_name='harry_potter_1' ,documents=splitted_harry, embedding=embedding, persist_directory="./Chroma")
```

```python
from pprint import pprint
query = "Who is the McGonagall?"
# res = collection.query(query_texts=[query], n_results=5)
res = vectorstore.similarity_search_with_score(query, k=5)
# res = [{'doc': doc, 'id': idx} for idx, (doc, id) in enumerate(zip(res['documents'][0], res['ids'][0]))]
res = [{'doc': doc[0].page_content, 'id': doc[0].id} for doc in res]
print(res)
```

```python
from langchain_core.prompts import ChatPromptTemplate
sys_prompt1 = """
You are a helpful retrieval and summarization agent, which gets queries from the user and some related information about that query and you must answer this query in accordance with the GIVEN INFORMATION ONLY.

Instructions and Formats:
- The query will be a simple text which you have to answer after looking into the related documents, DO NOT USE OTHER SOURCES OF INFORMATION, ONLY USE THE GIVEN RELATED DOCUMENTS TO FORM YOUR ANSWER.
- The related documents will be in the format where there will be the doc field which will contain the text information and an id field which will be an unique identifier to this particular doc.
- Whatever the documents are summarize their meaning in correspondance to the query itself and then form your output.

Query: {query}
Related Documents: {res}

Output Format:
[
  {{
    'content': "This is a text that has been refined and summarized by you and is sourced from the given information",
    'citation': <id>
  }},
  ...
]
In above format, there is a list of citied information, each element of the list is an object containing the content field which is the answer to the query along with the citation id of the sourced doc from the given related documents.
The size of the output list should be maximum of 3.


Rules:
- MAKE SURE THAT EACH STATEMENT IS SUPPORTED BY THE GIVEN FACTS.
- DO NOT USE ANOTHER KNOWLEDGE SOURCES OTHER THAN THAT IS PROVIDED.
- THE CITATION SHOULD ONLY MATCH THE DOCUMENT ID OF WHICH THE INFORMATION IS SOURCED FROM.
- IT SHOULD NOT BE THE CASE THAT THE ID OF A DOCUMENT IS CITIED TO THE OUTPUT WHICH DOESN'T HAVE THE SAME INFORMATION.
- IF THE RELATED INFORMATION IS NOT SUFFICENT TO ANSWER, JUST RETURN [ {{"content": "Insufficient information", "citation": [] }} ]
- OUTPUT SHOULD BE IN JSON STRING FORMAT WITH NO MARKDOWN FEATURES.
"""
sys_prompt1 = ChatPromptTemplate.from_template(sys_prompt1)
sys_prompt1 = sys_prompt1.invoke({'query': query, 'res': res})
```

```python
out = llm.invoke(sys_prompt1)
try: 
    out = json.loads(out.content)
    pprint(out)
except:
    print("The output wasn't in JSON format")
    print(out.content)
```
