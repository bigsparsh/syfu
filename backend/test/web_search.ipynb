```python
from ddgs import DDGS
from pprint import pprint

with DDGS() as ddgs:
	res = ddgs.text("Python programming", max_results=5)
	for r in res:
		print(f"Title: {r['title']}")
		print(f"Link: {r['href']}")
		pprint(f"Body: {r['body']}")
		print()
```

```python
import asyncio
from crawl4ai import AsyncWebCrawler

async def main():
    async with AsyncWebCrawler() as crawler:
        result = await crawler.arun(url="https://adityanandan.vercel.app/")
        print("Internal Links")
        for il in result.links['internal']:
            print(il['href'])
        print("External Links")
        for il in result.links['external']:
            print(il['href'])

await main()
```
