```python
from trafilatura import fetch_url, extract
import trafilatura
from markdownify import markdownify
import httpx
# downloaded = fetch_url('https://adityanandan.vercel.app/')
# extract(downloaded)
# print(extract(downloaded, output_format="json", with_metadata=True))
link = "https://adityanandan.vercel.app/"
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"
}
async with httpx.AsyncClient(headers=headers, follow_redirects=True, timeout=15.0) as client:
    resp = await client.get(link)
    html = resp.text
                    # Extract clean article markdown directly
    content = trafilatura.extract(
        html,
        url=link,
        output_format="markdown",
        favor_recall=True,
        include_tables=True,
        include_comments=True,
        include_links=True,
        include_images=True,
        include_formatting=True,
        with_metadata=True,
        deduplicate=False,
    )
    
    # 2. First fallback: trafilatura raw text extraction
    if not content or len(content.strip()) < 100:
        content = trafilatura.html2txt(html)
        
    # 3. Ultimate fallback: full page raw Markdown conversion
    if not content or len(content.strip()) < 100:
        content = markdownify(html, heading_style="ATX", strip=['script', 'style'])
    print(content)
# extract(downloaded, output_format="json", with_metadata=True)
```
